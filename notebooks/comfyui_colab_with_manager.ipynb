### Run ComfyUI with cloudflared (Recommended Way)




```python
#@title 🚀 Lancement de ComfyUI avec Cloudflare Stabilisé (Anti-403 & API OpenCode)

import os
import subprocess
import threading
import time
import socket
import sys

PORT = 8188
URL_FILE = "/content/comfyui_url.txt"

# 1. Téléchargement propre de Cloudflared
if not os.path.exists("/usr/local/bin/cloudflared"):
    print("📥 Téléchargement et installation de Cloudflared stable...")
    !wget -q -O /usr/local/bin/cloudflared https://github.com/cloudflare/cloudflared/releases/latest/download/cloudflared-linux-amd64
    !chmod +x /usr/local/bin/cloudflared
    print("✅ Cloudflared configuré avec succès!")

# 2. Thread de surveillance et de démarrage du tunnel Cloudflare
def iframe_thread(port):
    while True:
        time.sleep(1.0)
        sock = socket.socket(socket.AF_INET, socket.SOCK_STREAM)
        result = sock.connect_ex(('127.0.0.1', port))
        sock.close()
        if result == 0:
            break

    print("\n🔥 ComfyUI a démarré localement. Génération de votre lien Cloudflare...\n")

    # Tunnel vers localhost en écrasant l'hôte et l'origine pour contourner les verrous 403 d'aiohttp
    cmd = [
        "cloudflared", "tunnel",
        "--url", f"http://127.0.0.1:{port}",
        "--http-host-header", "localhost",
        "--origin-server-name", "localhost"
    ]
    p = subprocess.Popen(cmd, stdout=subprocess.PIPE, stderr=subprocess.STDOUT, text=True)

    for line in p.stdout:
        if "trycloudflare.com" in line:
            parts = line.split()
            for part in parts:
                if "trycloudflare.com" in part:
                    clean_url = part.strip().replace("http://", "https://")
                    print("\n" + "="*60)
                    print(f"🌍 CLIQUEZ ICI POUR ACCÉDER À COMFYUI : {clean_url}")
                    print("="*60 + "\n")

                    with open(URL_FILE, "w") as f:
                        f.write(clean_url)
                    break

# Lancement du tunnel en tâche de fond
threading.Thread(target=iframe_thread, daemon=True, args=(PORT,)).start()

# 3. Lancement de ComfyUI (Avec CORS activé pour autoriser l'accès externe via Cloudflare)
%cd $WORKSPACE
!python main.py --listen 0.0.0.0 --port 8188 --use-pytorch-cross-attention --enable-cors-header
```

### 🌐 Option : Lancement avec un lien unique permanent (Ngrok Static Domain)
Remplissez vos identifiants Ngrok ci-dessous pour obtenir un accès via un lien unique et persistant.

```python
#@title 🚀 Lancement permanent sécurisé avec Ngrok

import os
import threading
import time
import socket
from google.colab import userdata

PORT = 8188

# Récupération sécurisée des identifiants depuis le gestionnaire de secrets de Colab
try:
    NGROK_AUTHTOKEN = userdata.get('NGROK_AUTHTOKEN')
    NGROK_STATIC_DOMAIN = userdata.get('NGROK_STATIC_DOMAIN')
except Exception:
    NGROK_AUTHTOKEN = None
    NGROK_STATIC_DOMAIN = None

if not NGROK_AUTHTOKEN or not NGROK_STATIC_DOMAIN:
    print("❌ Erreur : Vos secrets 'NGROK_AUTHTOKEN' et 'NGROK_STATIC_DOMAIN' ne sont pas configurés dans l'onglet 🔑 Secrets de Google Colab.")
    print("Veuillez les ajouter avec les valeurs correspondantes et autoriser l'accès au notebook.")
else:
    # Installation de pyngrok
    !pip install -q pyngrok
    from pyngrok import ngrok

    # Configuration de l'authtoken
    ngrok.set_auth_token(NGROK_AUTHTOKEN)

    def launch_ngrok_tunnel(port):
        # Attente que le port local de ComfyUI devienne actif
        while True:
            time.sleep(1.0)
            sock = socket.socket(socket.AF_INET, socket.SOCK_STREAM)
            result = sock.connect_ex(('127.0.0.1', port))
            sock.close()
            if result == 0:
                break

        print("\n🔥 ComfyUI a démarré localement. Initialisation du tunnel permanent Ngrok...\n")
        try:
            # Fermer d'éventuels processus locaux ngrok résiduels
            ngrok.kill()

            # Lancement du tunnel avec le nom de domaine statique
            tunnel = ngrok.connect(port, proto="http", domain=NGROK_STATIC_DOMAIN)
            print("\n" + "="*60)
            print(f"🌍 VOTRE LIEN UNIQUE PERMANENT : {tunnel.public_url}")
            print("="*60 + "\n")
        except Exception as e:
            print(f"❌ Erreur lors de la configuration du tunnel Ngrok: {e}")

    # Lancement du thread Ngrok en tâche de fond
    threading.Thread(target=launch_ngrok_tunnel, daemon=True, args=(PORT,)).start()

    # Lancement de ComfyUI en local
    %cd $WORKSPACE
    !python main.py --listen 127.0.0.1 --port 8188 --use-pytorch-cross-attention
```

### Run ComfyUI with colab iframe (use only in case the previous way with localtunnel doesn't work)

You should see the ui appear in an iframe. If you get a 403 error, it's your firefox settings or an extension that's messing things up.

If you want to open it in another window use the link.

Note that some UI features like live image previews won't work because the colab iframe blocks websockets.

```python
import threading
import time
import socket
def iframe_thread(port):
  while True:
      time.sleep(0.5)
      sock = socket.socket(socket.AF_INET, socket.SOCK_STREAM)
      result = sock.connect_ex(('127.0.0.1', port))
      if result == 0:
        break
      sock.close()
  from google.colab import output
  output.serve_kernel_port_as_iframe(port, height=1024)
  print("to open it in a window you can open this link here:")
  output.serve_kernel_port_as_window(port)

threading.Thread(target=iframe_thread, daemon=True, args=(8188,)).start()

!python main.py --dont-print-server
```
